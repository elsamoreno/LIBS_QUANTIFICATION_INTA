```python
from scipy.interpolate import interp1d
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import numpy as np

def get_interpolated_normalized(P, wave_common):
    interp_func = interp1d(P['Wave'], P['Sample'] / np.max(P['Sample']), kind='linear')
    return interp_func(wave_common)

# Crear eje común global
wave_min = max(P1_T1['Wave'].min(), P2_T1['Wave'].min(), P3_T1['Wave'].min(),
               P1_T2['Wave'].min(), P2_T2['Wave'].min(), P3_T2['Wave'].min(),
               P1_T3['Wave'].min(), P2_T3['Wave'].min())
wave_max = min(P1_T1['Wave'].max(), P2_T1['Wave'].max(), P3_T1['Wave'].max(),
               P1_T2['Wave'].max(), P2_T2['Wave'].max(), P3_T2['Wave'].max(),
               P1_T3['Wave'].max(), P2_T3['Wave'].max())

wave_common = np.linspace(wave_min, wave_max, 1000)

# Interpolar y normalizar todos los espectros
X_all = []
labels = []
colors = []

for label, color, group in zip(['T1', 'T2', 'T3'], ['red', 'green', 'blue'],
                               [(P1_T1, P2_T1, P3_T1), (P1_T2, P2_T2, P3_T2), (P1_T3, P2_T3)]):
    for i, P in enumerate(group):
        X_all.append(get_interpolated_normalized(P, wave_common))
        labels.append(f"{label} - Pos {i+1}")
        colors.append(color)

# Convertir a array y aplicar PCA
X_all = np.array(X_all)
X_scaled = StandardScaler().fit_transform(X_all)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Graficar
plt.figure(figsize=(8,6))
for i in range(len(X_pca)):
    plt.scatter(X_pca[i, 0], X_pca[i, 1], color=colors[i], label=labels[i])
    plt.annotate(labels[i], (X_pca[i, 0]+0.3, X_pca[i, 1]+0.3), fontsize=9)

plt.xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)')
plt.ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)')
plt.title('PCA combinada de T1, T2 y T3')
plt.grid(True)
# Evitar etiquetas duplicadas en la leyenda
handles, labels_unique = plt.gca().get_legend_handles_labels()
by_label = dict(zip(labels_unique, handles))
plt.legend(by_label.values(), by_label.keys())
plt.tight_layout()
plt.show()
```
